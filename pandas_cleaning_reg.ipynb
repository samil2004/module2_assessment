```python
df = df.drop_duplicates()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df
```

```python
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
df
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
df
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```
